#### Load Env Variables


```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
res = llm.invoke("python program to print hello world")
print(res.text)
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4o")
res = llm.invoke("types of GenAI")
print(res.content)
```

Static Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")

prompts = [                                    #list of tuples
    ("system", "You are a Python Developer"),  #tuples
    ("user", "Write a program prints 1 to 10")
]

res = llm.invoke(prompts)
print(res.text)
```

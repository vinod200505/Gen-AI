```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", streaming = True)

questions = "Who is Virat Kohli and his profession ?"

res = llm.stream(questions)

for chunk in res:
    print(chunk.text, end="")

```

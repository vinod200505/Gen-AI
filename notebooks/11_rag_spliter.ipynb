```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
texts = "Good morning, respected teachers and my dear friends \
        Today, I would like to speak about an important topic: **“Easy for Farmers, Better for Farming.”** \
        Farmers play a very important role in our lives. They work hard every day to provide food for us. \
        But they face many challenges, such as unpredictable weather, high farming costs, low crop prices, \
        and difficulties in reaching markets."
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 80, chunk_overlap = 20)
```

```python
chunks = splitter.split_text(texts)
chunks
```

### Document Text Loader

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/speech.txt", encoding="utf-8")
texts = loader.load()
```

```python
pdfloader = PyPDFLoader("../data/1.Generative AI course.pdf")
pdfs = pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
```

```python
text_chunks = splitter.split_documents(texts)
len(text_chunks)
```

```python
pdf_chunks = splitter.split_documents(pdfs)
len(pdf_chunks)
```

```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name="Vinod Rathod", age="21")
user
```

### Field Validation

```python
class Product(BaseModel):
    title : str = Field(max_length=10)
    des : str = Field(min_length=20)
    
    price : int = Field(gt=0, le=100)
```

```python
product = Product(title="Samsung", des="Samung is best mobile in the world", price=10)

product
```

```python
class Order(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    product:Product
```

```python
order = Order(id="ndkjg74", product=product)

order.product.title
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: List[Product]
    comments: str = Field(
        default="No Comments"
    )
```

```python
cart = Carts(id="nen48njw", items=[product], comments="This is my cart")

cart
```

#### Annotated

```python
commonValidator = Annotated[List[Product], Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: commonValidator
    comments: str = Field(
        default="No Comment"
    )
```

```python
class Order(BaseModel):
    products: commonValidator
```

```python
ord = Order(products=[product])

ord
```

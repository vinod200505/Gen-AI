```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import GoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = GoogleGenerativeAI(model="gemini-3.5-flash")

out = StrOutputParser()

def tranform_parse(result: str):
    return result.upper()

prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "you are a translator and translate input in {language}"},
    {"role": "user", "content": "{query}"}
])

chain = prompts | llm | out | tranform_parse

res = chain.invoke({"language": "French", "query": "Hello, how are you?"})

print(res)
```
